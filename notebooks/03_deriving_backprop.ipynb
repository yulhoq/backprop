## 1. What is Gradient Descent?

Gradient descent is a widely used optimization algorithm that is commonly used to train machine learning models and neural networks. It trains machine learning models to find optimal weights and bias by minimizing errors between predicted and actual results.

Think of it like walking down a hill in fog.
You cannot see the bottom, but you can feel the slope.

## 2. Loss Function Intuition

Think of loss like a **mistake meter**.

- If prediction is perfect → loss = 0  
- If prediction is  very wrong → loss is large  

So, loss is a number that tells you that how bad the model is performing.

A loss function helps us to find out the model's loss by comparing the model's prediction with actual values.


$L(\theta) = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

- yᵢ = actual (true) value  
- ŷᵢ = predicted value from the model  
- n = number of data points  
- L(θ) = total loss (error of the model)

The entire training process of a neural network is

👉 Find values of weights & biases that minimize Loss.

```python
# let's run a loss fuction on this data to see how wrong our prediction to actual data
Epoch 200, Loss = 99396.3474

loss = np.mean((y - y_pred) ** 2)

loss
```

## 3, Gradient

Now know that how loss calcute from **forwoard pass**. But how much each parameter (e.g. weight, bias) contributed to errors, and in which direction  then will go to reach the downhill?

To answer this, we compute the gradient of the loss with respect to each parameter.
The gradient gives us two important pieces of information:

1. The strength of the influence of a particular parameter on the loss.
2. Which way the parameter needs to be changed in order to minimize the loss.

Based on these pieces of information, Gradient Descent adjusts the parameters accordingly, making sure they take small steps to minimize the loss.

In simple terms, the gradient shows the way to the optimal point, where the loss is minimized.

```python
n = len(x)

d_w = (-2 / n) * np.sum(x * (y - y_pred))
d_b = (-2 / n) * np.sum(y - y_pred)

print("dw =", d_w)
print("db =", d_b)
```

## 4. Learning Rate

Here, α is the learning rate.

It determines the extent of each update to the model parameters and the speed of learning from errors of a model, whether it is fast or slow.

```python
# determinr the learning rateEpoch 200, Loss = 99396.3474


learning_rate = 0.0001
```

## 5. Core Update Rule
 
θ=θ−α∇L(θ)

Where:

- θ = model parameters (weights and biases)
- α  = learning rate
- ∇L(θ) = gradient of the loss function with respect to the parameters

The gradient is the direction of steepest increase in the loss. We want to minimize the loss so we go in the opposite direction of the gradient.

This updating process is repeated during training until the model reaches a minimum loss.

```python
# updation of parameters

w = w - d_w * learning_rate
b = b - d_b * learning_rate

print("updated w: ", w)
print("updated b: ", b)
```

```python
# now let's see if ouor model improve

y_pred = w * x + b

new_loss = np.mean((y - y_pred) ** 2)

old_loss = loss
print("old loss: ", old_loss)
print("new loss: ", new_loss)
```

```python
# the whole training loop 

x = np.array([2,5,6,3,7])
y = np.array([4,7,-9,2,4])
w = 0.5
b = 0.0

learning_rate = 0.0000001
epochs = 1000

n = len(x)

for epoch in range(epochs):

    # Forward pass
    y_pred = w * x + b

    # Loss
    loss = np.mean((y - y_pred) ** 2)

    # Gradients
    dw = (-2 / n) * np.sum( x* (y - y_pred))Epoch 200, Loss = 99396.3474

    db = (-2 / n) * np.sum(y - y_pred)

    # Gradient Descent update
    w = w - learning_rate * d_w
    b = b - learning_rate * d_b

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.4f}")
```

## 7. Connection to Backpropagation


When training the neural network, Gradient Descent and Backpropagation start working together.

Backpropagationcalculates gradients of the loss function with respect to model parameters.
 
Then, Gradient Descent makes use of those gradients to update the weights and biases.

It means that
Backpropagation provides the model with instructions for its self-modification.
- This is how Gradient Descent does it.
