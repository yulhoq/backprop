## How Do We Learn?
Think about how human brain works: 

It recognizes the pattern that we see through our eyes. It matches this pattern with our brain's pre-recognizing things. That's how the brain signals to us that "yeah, it is the cat" because our brain knows what a cat looks like. We simply compare what we see with the patterns our brain has previously recognized.


## What is Neural Network?

Neural network is a mathematical model that is inspired by human brain. In this case, it learns from data by adjusting "weights" over and over. This allows a neural network to learn rules on its own. 

So, we see a biological brain learn = by changing synapse strength.
On the other hand, an artificial neuron learns = by adjusting weights + gradient descent.

## The Most Simple Neuron


The simplest neuron is a weighted sum of inputs.

Output formula:

y = w₁x₁ + w₂x₂ + ... + wₙxₙ

Here, w's are weights and x's are input values.

```python
# inputs
x_1 = 10
x_2 = 19

# weights
w_1 = 3
w_2 = 4

# formula 
y = w_1*x_1 + w_2*x_2 

print("Output = ", y)
```

## 3. Example

Imagine a neuron predicting exam success.

Inputs:

Study hours = 4  
Sleep hours = 7  

Weights:

Study weight = 0.8  
Sleep weight = 0.2  

Output:

y = (4 × 0.8) + (7 × 0.2)

```python
# inputs 
study_hours = 4
sleep_hours = 7

# weights
study_weight = 0.8
sleep_weight = 0.2

# output 
y = (study_hours * study_weight) + (sleep_hours * sleep_weight)
print("Output =", y)
```
