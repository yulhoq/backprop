## What is Forward Propagation?

**Forward propagation** is the output of 1st estimation of a neural network using the training data. It is the process when a neural network estimates the predictions or outputs using current weights and biases where our training data, which passes through a neural network layer by layer.


In a big neural network, where numerous layers exist, each neuron gets the information from the previous layers and perform calculation using weights and biases, applies activation function and pass it the next layer. 


**Mathematical Foundation for a single neuron**

The main formula is

z = W_T x + b

here, 
x and w both are vectors. 

Initially, w vector is taken randomly and can be a zero vector or anything a programmer likes. 
 
Inputs: x = [2, 5, -1]
Weights: w = [0.5, -1, 2]
Bias: b = 0.5

z = (x_1 * w_1) + (x_2 * w_2) + (x_3 * w_3) + b

z = (2 * 0.5) + (5 * -1) + (-1 * 2) + 0.5 = -5.5

```python
import numpy as np

x = [2, 5, -1]
w = [0.5, -1, 2]
b = 0.5

z = np.dot(x, w) + b

print("Output: ", z)
```

## Next we apply a activation funtion

Without activation functions, the network would behave like a single linear model regardless of depth.

The sigmoid function is defined as $\sigma(x) = \frac{1}{1 + e^{-x}}$.

And that create a prediction.

```python
def sigmoid(z):
    y = 1/(1 - np.exp(-z))
    return y
```

```python
y_hat = sigmoid(z)

print("Y_hat =", y_hat)
```

## Flow of Information

In a deep neural network:

- Input data is passed into the first layer.
- Each neuron computes a weighted sum of inputs.
- An activation function is applied.
- The output is passed to the next layer.
